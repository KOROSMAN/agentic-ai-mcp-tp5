```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen3:4b",
    temperature=0,      # réponses stables, sans fantaisie
    reasoning=True,    # qwen3 « réfléchit » à voix haute par défaut : on masque ce raisonnement
)

reponse = llm.invoke("Bonjour, qui es-tu ? Réponds en une phrase.")
print(reponse.content)
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

agent = create_agent(
    model=llm,
    system_prompt="Tu es un assistant utile. Réponds en français, brièvement.",
)

resultat = agent.invoke({"messages": [HumanMessage("Explique ce qu'est un agent IA en 2 phrases.")]})
print(resultat["messages"][-1].content)
```

```python
for message in resultat["messages"]:
    print(f"{type(message).__name__:15s} | {message.content[:100]}")
```

```python
agent.invoke({"messages": [HumanMessage("Je m'appelle Ali.")]})

resultat = agent.invoke({"messages": [HumanMessage("Comment je m'appelle ?")]})
print(resultat["messages"][-1].content)
```

```python
from langgraph.checkpoint.memory import InMemorySaver

agent_memoire = create_agent(
    model=llm,
    system_prompt="Tu es un assistant utile. Réponds en français, brièvement.",
    checkpointer=InMemorySaver(),   # la mémoire de l'agent
)

config = {"configurable": {"thread_id": "conversation-1"}}

agent_memoire.invoke({"messages": [HumanMessage("Je m'appelle Ali.")]}, config=config)
resultat = agent_memoire.invoke({"messages": [HumanMessage("Comment je m'appelle ?")]}, config=config)
print(resultat["messages"][-1].content)
```

```python
autre_config = {"configurable": {"thread_id": "conversation-2"}}

resultat = agent_memoire.invoke({"messages": [HumanMessage("Comment je m'appelle ?")]}, config=autre_config)
print(resultat["messages"][-1].content)

# Retour sur la première conversation : l'historique complet est toujours là
etat = agent_memoire.get_state(config)
print("\nMessages mémorisés dans conversation-1 :", len(etat.values["messages"]))
```
